# **POSTWORK 02 M2**

Imagina la siguiente situación: trabajas en una empresa de desarrollo de software y tu jefe te ha pedido que diseñes un programa para gestionar el estado de los proyectos en curso. Este programa será fundamental para llevar un seguimiento adecuado de los proyectos y asegurarse de que se mantengan en los plazos establecidos.

Tu tarea es desarrollar un programa en Python que permita a los usuarios registrar y actualizar el estado de varios proyectos. Los proyectos pueden estar en diferentes estados: no iniciado, en progreso y completado. El programa permitirá actualizar y mostrar el estado de los proyectos, y permanecerá en ejecución hasta que el usuario decida salir

El programa debe realizar las siguientes acciones:

1. Solicitará al usuario seleccione que una opción: registrar nuevo proyecto, actualizar estado de proyecto, mostrar lista de proyectos o salir.
2. Dependiendo la opción, solicitará al usuario el nombre del proyecto.
3. Pedirá al usuario que ingrese el estado del proyecto (no iniciado, en progreso, completado).
4. Permitirá al usuario actualizar el estado de los proyectos registrados.
5. Mostrará una lista de todos los proyectos con su estado actual.
6. Finalmente, el programa debe mostrar un mensaje con el nombre y el estado de todos los proyectos registrados cuando el usuario decida mostrar la lista de proyectos.

Para ello, el programa debe seguir estos pasos:
  
  • Registrar el nombre del proyecto. Solicitar al usuario que ingrese el nombre del proyecto que desea registrar.
  
  • Solicitar el estado del proyecto. Pedir al usuario que ingrese el estado actual del proyecto. Los estados pueden ser "no iniciado", "en progreso" o "completado".

  • Actualizar el estado del proyecto. Permitir que el usuario actualice el estado de un proyecto existente.

  • Mostrar la lista de proyectos. Utilizar un bucle para iterar a través de la lista de proyectos y mostrar su estado actual.

  • Ejecución continua. Mantener el programa en ejecución hasta que el usuario decida salir, permitiendo múltiples operaciones.

  Requisitos adicionales:
  
  ✓ Interacción con el usuario. El programa debe utilizar la función input() para leer los datos del usuario.
  
   ✓ Condicionales. Utilizar sentencias if y match para manejar las diferentes acciones y estados de los proyectos.
   
   ✓ Bucles. Utilizar while para mantener el programa en ejecución
   
   ✓ Presentación de resultados. Utilizar la interpolación de strings para mostrar mensajes claros y dinámicos al usuario.

```python
#Se crea un diccionario para guardar y actualizar información

proyectos = {}
edos_proyectos_permitidos = ["no iniciado", "en progreso", "completado"]
programa = True

#Utilizamos un bucle
while programa == True:
  print("\n*** MENU DE OPCIONES ***")
  print("1. Registrar nuevo proyecto")
  print("2. Actualizar estado del proyecto")
  print("3. Mostrar lista de proyectos registrados")
  print("4. Salir")

  opcion = input("Elige una de las opciones: ")

  match opcion:
    case "1":
      nombre_proyecto = input("\nIngresa el nombre del proyecto: ").strip().lower()
      print(f"\nEstados permitidos: {', '.join(edos_proyectos_permitidos)}")
      edo_proyecto = input("Ingresa el estado en el que se encuentra el proyecto: ").lower()
      #Vamos a guardar en el diccionario
      proyectos[nombre_proyecto] = edo_proyecto
      print(f"\nEl proyecto '{nombre_proyecto}' se registro correctamente.")

    case "2":
      nombre_proyecto = input("\nIngresa el nombre del proyecto que deseas actualizar: ").strip().lower()
      # Verificamos si el proyecto ya fue guardado antes
      if nombre_proyecto in proyectos:
        print(f"Estado actual de '{nombre_proyecto}': {proyectos[nombre_proyecto]}")
        print(f"\nEstados permitidos: {', '.join(edos_proyectos_permitidos)}")
        nuevo_estado = input("Ingresa el nuevo estado: ").strip().lower()
        if nuevo_estado in edos_proyectos_permitidos:
          proyectos[nombre_proyecto] = nuevo_estado
          print(f"El estado del proyecto '{nombre_proyecto}' se ha actualizado a '{nuevo_estado}' exitosamente.")
        else:
          print(f"Estado inválido. No se realizó ningún cambio.")
          nuevo_estado = input("\nIngresa el nuevo estado: ").strip().lower()
          proyectos[nombre_proyecto] = nuevo_estado
          print(f"\nEstado del proyecto '{nombre_proyecto}' actualizado correctamente.")
      else:
        print(f"\nEl proyecto '{nombre_proyecto}' no existe.")

    case "3":
      print("\n*** LISTA DE PROYECTOS REGISTRADOS ***")
      # Verificamos si el diccionario tiene proyectos guardados
      if len(proyectos) > 0:
        for nombre, estado in proyectos.items():
          print(f"• Proyecto: {nombre} | Estado: {estado}")
      else:
        print("\nNo hay proyectos registrados hasta el momento.")

    case "4":
      print("\nSaliendo del programa...")
      break

    case _:
      print(f"\nOpción inválida. Por favor, ingresa un número del menú del 1-4")

  if programa and opcion in ["1", "2", "3"]:
    continuar = input("\n¿Deseas realizar otra acción? (si/no): ").strip().lower()
    if continuar not in ["s", "si", "sí"]:
            print("\nFinalizando el programa...")
            programa = False
```
